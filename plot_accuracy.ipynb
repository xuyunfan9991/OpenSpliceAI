```python
import matplotlib.pyplot as plt
import numpy as np

# 读取准确率数据
file_path = '/home1/xyf/data/openspliceai_tissue_data/train_model/reference_model_400/SpliceAI_reference_model_400_focal_loss0203_400_0_rs42/0/LOG/TEST/accuracy.txt'

epochs = []
accuracies = []

with open(file_path, 'r') as f:
    for line in f:
        line = line.strip()
        if line:
            accuracies.append(float(line))

# epochs从0开始
epochs = list(range(len(accuracies)))

# 绘制折线图
plt.figure(figsize=(12, 6))
plt.plot(epochs, accuracies, marker='o', linestyle='-', linewidth=2, markersize=4)
plt.xlabel('Epoch', fontsize=12)
plt.ylabel('Accuracy', fontsize=12)
plt.title('Test Accuracy per Epoch', fontsize=14, fontweight='bold')
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

# 打印统计信息
print(f"总epoch数: {len(accuracies)}")
print(f"最高准确率: {max(accuracies):.4f} (Epoch {accuracies.index(max(accuracies))})")
print(f"最低准确率: {min(accuracies):.4f} (Epoch {accuracies.index(min(accuracies))})")
print(f"平均准确率: {np.mean(accuracies):.4f}")
```

```python
# 读取acceptor F1分数数据
file_path_f1 = '/home1/xyf/data/openspliceai_tissue_data/train_model/reference_model_400/SpliceAI_reference_model_400_focal_loss0203_400_0_rs42/0/LOG/TEST/acceptor_f1.txt'

epochs_f1 = []
f1_scores = []

with open(file_path_f1, 'r') as f:
    for line in f:
        line = line.strip()
        if line:
            f1_scores.append(float(line))

# epochs从0开始
epochs_f1 = list(range(len(f1_scores)))

# 绘制折线图
plt.figure(figsize=(12, 6))
plt.plot(epochs_f1, f1_scores, marker='o', linestyle='-', linewidth=2, markersize=4, color='green')
plt.xlabel('Epoch', fontsize=12)
plt.ylabel('Acceptor F1 Score', fontsize=12)
plt.title('Test Acceptor F1 Score per Epoch', fontsize=14, fontweight='bold')
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

# 打印统计信息
print(f"总epoch数: {len(f1_scores)}")
print(f"最高F1分数: {max(f1_scores):.4f} (Epoch {f1_scores.index(max(f1_scores))})")
print(f"最低F1分数: {min(f1_scores):.4f} (Epoch {f1_scores.index(min(f1_scores))})")
print(f"平均F1分数: {np.mean(f1_scores):.4f}")
```
